```python
# Case Folding in NLP

# Case folding converts text to lowercase.
# This makes "Hello", "HELLO", and "hello" treated as the SAME word.

text = "Hello WORLD"
print(text.lower())
```

```python
# Example 1: Why case folding matters

# Without case folding: 3 different words
words = ["hello", "Hello", "HELLO"]
unique_no_fold = len(set(words))
print("Without folding:", unique_no_fold, "unique words")
print(words)

# With case folding: 1 word
folded = [w.lower() for w in words]
unique_with_fold = len(set(folded))
print("\nWith folding:", unique_with_fold, "unique word")
print(folded)
```

```python
# Example 2: Simple comparison

# Before: "hello" != "Hello"
print("hello" == "Hello")

# After: both become "hello"
print("hello".lower() == "Hello".lower())
```

```python
# Example 3: Counting words

text = "Python python PYTHON nlp NLP"

# Without folding
words = text.split()
print("Without folding:", len(set(words)), "unique")
print(set(words))

# With folding
folded = [w.lower() for w in words]
print("\nWith folding:", len(set(folded)), "unique")
print(set(folded))
```

```python
# Example 4: Whole sentence

sentence = "The Quick BROWN Fox jumped OVER the lazy dog"

print("Before:")
print(sentence)

print("\nAfter:")
print(sentence.lower())
```

```python
# Example 5: When to use case folding

# Use it for: email, search, filtering
# Don't use it for: brand names, author ID, sentiment

text1 = "URGENT: Read this"      # Shouting loses meaning
text2 = "Apple Inc."              # Brand name loses identity

print("Original:", text1)
print("Folded:  ", text1.lower())
print()
print("Original:", text2)
print("Folded:  ", text2.lower())
```

```python
# Summary
# 1. .lower() turns all text to lowercase
# 2. Makes "Hello", "HELLO", "hello" the same
# 3. Reduces different versions of same word
# 4. Use when case doesn't matter for your task
```
