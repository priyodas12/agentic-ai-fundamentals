```python
# Special Character Removal in NLP

# Remove punctuation and symbols like: ! @ # $ % & ? .
# This makes "Hello!" and "Hello?" become the SAME word "Hello"

import re

text = "Hello! How are you?"
clean = re.sub(r"[^a-zA-Z0-9\s]", "", text)

print("Before:", text)
print("After: ", clean)
```

```python
# Example 1: Remove simple punctuation

texts = [
    "Hello!",
    "Hi?",
    "Good.",
    "Yes!!!"
]

for t in texts:
    clean = re.sub(r"[^a-zA-Z0-9\s]", "", t)
    print(f"{t:<15} -> {clean}")
```

```python
# Example 2: Remove symbols and special chars

emails = [
    "user@example.com",
    "price: $99",
    "hashtag #nlp",
    "email@test.org"
]

for e in emails:
    clean = re.sub(r"[^a-zA-Z0-9\s]", "", e)
    print(f"{e:<25} -> {clean}")
```

```python
# Example 3: Lowercase + remove chars

text = "HELLO! WORLD?"

# Step 1: lowercase
step1 = text.lower()
print("Step 1 (lowercase):", step1)

# Step 2: remove special chars
step2 = re.sub(r"[^a-z0-9\s]", "", step1)
print("Step 2 (clean):   ", step2)
```

```python
# Example 4: Real sentences

sentences = [
    "How are you?",
    "I'm happy!",
    "Email: john@test.com",
    "Price is $50!"
]

print("BEFORE -> AFTER")
print("-" * 40)
for s in sentences:
    clean = re.sub(r"[^a-zA-Z0-9\s]", "", s)
    print(f"{s:<25} -> {clean}")
```

```python
# Example 5: Create a cleaning function

def clean(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    return text

texts = [
    "Hello World!",
    "Test@123",
    "Python 3.11!"
]

for t in texts:
    print(f"{t:<20} -> {clean(t)}")
```

```python
# Example 6: Different patterns

text = "Python 3.11 (2022) - $30.50"

# Keep everything
print("Keep all:        ", text)

# Remove only punctuation
pattern1 = re.sub(r"[!?.]", "", text)
print("No . ! ?:        ", pattern1)

# Remove everything but letters
pattern2 = re.sub(r"[^a-zA-Z\s]", "", text)
print("Only letters:    ", pattern2)

# Remove everything but alphanumeric
pattern3 = re.sub(r"[^a-zA-Z0-9\s]", "", text)
print("Letters + numbers:", pattern3)
```

```python
# Summary
# 1. Special chars: ! ? . @ # $ % & etc
# 2. regex: [^a-zA-Z0-9\s] means "remove anything that's not letter/number/space"
# 3. Usually do: lowercase first, then remove chars
# 4. Use for cleaning messy text (emails, social media, web scraped data)
```
