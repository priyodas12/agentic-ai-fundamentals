```python
# Contractions in NLP

# A contraction is a shortened word: can't, don't, it's
# We expand them to: cannot, do not, it is
# This makes "can't" and "cannot" become the SAME word

import contractions

text = "I can't believe it"
expanded = contractions.fix(text)

print("Before:", text)
print("After: ", expanded)
```

```python
# Example 1: Common contractions

contractions_list = [
    "can't",
    "don't",
    "won't",
    "it's",
    "I'm",
    "you're"
]

print("CONTRACTION -> EXPANDED")
for c in contractions_list:
    expanded = contractions.fix(c)
    print(f"{c:<15} -> {expanded}")
```

```python
# Example 2: Simple sentence

text = "I can't do it"
expanded = contractions.fix(text)

print("Before:", text)
print("After: ", expanded)
```

```python
# Example 3: Multiple contractions

text = "They're going. She's happy. We won't stay."
expanded = contractions.fix(text)

print("Before:")
print(text)
print("\nAfter:")
print(expanded)
```

```python
# Example 4: Why it matters

from nltk.tokenize import word_tokenize

text = "can't go"

# Without expanding
words_before = word_tokenize(text)
print("Without expand:", words_before)

# After expanding
text_expanded = contractions.fix(text)
words_after = word_tokenize(text_expanded)
print("After expand:  ", words_after)

# "can't" -> "ca" + "n't" (bad!)
# "cannot" -> "can" + "not" (good!)
```

```python
# Example 5: Real text

texts = [
    "I'm happy",
    "They've won",
    "She'd like it",
    "It's raining"
]

for t in texts:
    exp = contractions.fix(t)
    print(f"{t:<20} -> {exp}")
```

```python
# Summary
# 1. Contractions: short forms like can't, don't, it's
# 2. Expand them to full forms: cannot, do not, it is
# 3. Makes different versions of same word become one word
# 4. Do this BEFORE tokenization for best results
```
