```python
# Tokenization in NLP

# Tokenization splits text into smaller pieces called TOKENS
# Usually: words, punctuation, or sentences

from nltk.tokenize import word_tokenize, sent_tokenize

text = "Hello world. How are you?"

print("WORD TOKENS:")
print(word_tokenize(text))

print("\nSENTENCE TOKENS:")
print(sent_tokenize(text))
```

```python
# Example 1: Word tokenization

text = "I like cats"

tokens = word_tokenize(text)
print("Text:  ", text)
print("Tokens:", tokens)
```

```python
# Example 2: Sentence tokenization

text = "Hello world. How are you? Great!"

sentences = sent_tokenize(text)
print("Text:     ", text)
print("Sentences:")
for s in sentences:
    print(" -", s)
```

```python
# Example 3: Punctuation as tokens

text = "Hi! How are you?"

tokens = word_tokenize(text)
print("Text:  ", text)
print("Tokens:", tokens)

# Notice: punctuation is separate
# "Hi" and "!" are two different tokens
```

```python
# Example 4: Why not use .split()?

text = "Hello, world!"

# Using .split()
split_way = text.split()
print("Using .split():  ", split_way)
# Result: ["Hello,", "world!"]  <- punctuation stuck to words

# Using word_tokenize()
tokenize_way = word_tokenize(text)
print("Using tokenize:  ", tokenize_way)
# Result: ["Hello", ",", "world", "!"]  <- punctuation separate
```

```python
# Example 5: Real example

text = "Dr. Smith's clinic costs $50."

tokens = word_tokenize(text)
print("Text:  ", text)
print("Tokens:", tokens)

# Notice:
# - "Dr." stays together
# - "Smith's" becomes "Smith" + "'s"
# - "$50" becomes "$" + "50"
# - "." is separate
```

```python
# Example 6: Different tokenizers

from nltk.tokenize import word_tokenize, wordpunct_tokenize
import re

text = "don't stop"

print("word_tokenize:  ", word_tokenize(text))
print("wordpunct:      ", wordpunct_tokenize(text))

# Different tokenizers split differently!
```

```python
# Summary
# 1. Tokenization: split text into small pieces (tokens)
# 2. word_tokenize: splits into words + punctuation
# 3. sent_tokenize: splits into sentences
# 4. Punctuation becomes separate tokens
# 5. Different tokenizers have different rules
```
