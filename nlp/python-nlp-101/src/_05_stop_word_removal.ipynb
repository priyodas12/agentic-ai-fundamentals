```python
# Stop Word Removal in NLP

# Stop words are common words like: the, is, a, and, or, it
# These words add little meaning. We remove them to focus on important words.

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "This is a sample sentence"
tokens = word_tokenize(text)

print("Before removing stop words:")
print(tokens)

# Remove stop words
stop_words = stopwords.words('english')
clean = [w for w in tokens if w not in stop_words]

print("\nAfter removing stop words:")
print(clean)
```

```python
# Example 1: What are stop words?

from nltk.corpus import stopwords

stop_words = stopwords.words('english')

print(f"Total stop words: {len(stop_words)}\n")
print("First 20 stop words:")
print(stop_words[:20])

# Common ones: the, is, a, and, or, it, to, in, of, that, this, etc
```

```python
# Example 2: Simple removal

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = "the dog is running"
tokens = word_tokenize(text)

stop_words = stopwords.words('english')
clean = [w for w in tokens if w not in stop_words]

print("Original:", text)
print("Tokens:  ", tokens)
print("Clean:   ", clean)
```

```python
# Example 3: Sentence before and after

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = "I am learning Python and I love it"

tokens = word_tokenize(text)
stop_words = stopwords.words('english')
clean = [w for w in tokens if w not in stop_words]

print("BEFORE:")
print(text)
print("\nAFTER (stop words removed):")
print(' '.join(clean))
```

```python
# Example 4: Real text example

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = "The quick brown fox jumps over the lazy dog"

tokens = word_tokenize(text)
stop_words = stopwords.words('english')
clean = [w for w in tokens if w not in stop_words]

print("Original:", ' '.join(tokens))
print("Cleaned: ", ' '.join(clean))

# "The", "the", "over" removed because they're common words
```

```python
# Example 5: Multiple sentences

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords

text = "I am happy. You are happy. We are all happy."

stop_words = stopwords.words('english')

print("BEFORE:")
print(text)

print("\nAFTER:")
sentences = sent_tokenize(text)
for s in sentences:
    tokens = word_tokenize(s)
    clean = [w for w in tokens if w not in stop_words]
    print(' '.join(clean))
```

```python
# Example 6: Create a cleaning function

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

def remove_stop_words(text):
    tokens = word_tokenize(text)
    stop_words = stopwords.words('english')
    clean = [w for w in tokens if w not in stop_words]
    return ' '.join(clean)

texts = [
    "the dog is running",
    "I love to code",
    "this is great"
]

for t in texts:
    print(f"{t:<30} -> {remove_stop_words(t)}")
```

```python
# Example 7: When NOT to remove stop words

# Good to remove: email filtering, search keywords
# Bad to remove: sentiment analysis, machine translation

text1 = "I love this movie"   # "love" = positive. Keep it!
text2 = "I hate this movie"   # "hate" = negative. Keep it!

print("Without removing 'I', 'this':")
print("Sentence 1:", text1)
print("Sentence 2:", text2)
print("Both have same meaning if we remove stop words!\n")

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

tokens1 = word_tokenize(text1.lower())
tokens2 = word_tokenize(text2.lower())

stop_words = stopwords.words('english')
clean1 = [w for w in tokens1 if w not in stop_words and w.isalpha()]
clean2 = [w for w in tokens2 if w not in stop_words and w.isalpha()]

print("After removing stop words:")
print("Cleaned 1:", clean1)
print("Cleaned 2:", clean2)
print("They look the same but mean opposite things!")
```

```python
# Summary
# 1. Stop words: common words like "the", "is", "a", "and"
# 2. Removing them reduces noise and text size
# 3. Good for: search, keywords, topic modeling
# 4. Bad for: sentiment analysis, translation, grammar
# 5. Use when the exact words don't matter, only meaning
```
