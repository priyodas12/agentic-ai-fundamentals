```python
# N-Grams in NLP

# An N-gram is a group of N words together.
# Examples:
#   1-gram (unigram): "I", "love", "Python"
#   2-gram (bigram): "I love", "love Python"
#   3-gram (trigram): "I love Python"

from nltk import ngrams
from nltk.tokenize import word_tokenize

text = "I love Python"
tokens = word_tokenize(text)

print("Tokens:", tokens)
print("\n1-grams (unigrams):")
print(list(ngrams(tokens, 1)))

print("\n2-grams (bigrams):")
print(list(ngrams(tokens, 2)))

print("\n3-grams (trigrams):")
print(list(ngrams(tokens, 3)))
```

```python
# Example 1: Bigrams (2-grams)

from nltk import ngrams
from nltk.tokenize import word_tokenize

text = "the cat sat on the mat"
tokens = word_tokenize(text)

bigrams = list(ngrams(tokens, 2))

print("Text:   ", text)
print("Tokens: ", tokens)
print("Bigrams:")
for b in bigrams:
    print(" -", b)
```

```python
# Example 2: Different n values

from nltk import ngrams
from nltk.tokenize import word_tokenize

text = "I love machine learning"
tokens = word_tokenize(text)

print("Text:  ", text)
print("Tokens:", tokens)

for n in [1, 2, 3]:
    n_grams = list(ngrams(tokens, n))
    print(f"\n{n}-grams:")
    print(n_grams)
```

```python
# Example 3: Create bigrams manually

text = "hello world python"
words = text.split()

print("Words:", words)

print("\nBigrams (manual):")
for i in range(len(words) - 1):
    bigram = (words[i], words[i+1])
    print(" -", bigram)
```

```python
# Example 4: Count bigrams

from nltk import ngrams
from nltk.tokenize import word_tokenize
from collections import Counter

text = "the cat sat on the mat and the cat jumped"
tokens = word_tokenize(text)

bigrams = list(ngrams(tokens, 2))
counts = Counter(bigrams)

print("Text:   ", text)
print("\nBigram counts:")
for bigram, count in counts.most_common(5):
    print(f"  {bigram}: {count}")
```

```python
# Example 5: Why use n-grams?

# Unigrams: lose word order, meaning
# Bigrams: keep some word order and context

from nltk import ngrams
from nltk.tokenize import word_tokenize

text = "I love cats I love dogs"
tokens = word_tokenize(text)

unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))

print("Text:", text)
print("\nUnigrams:", [u[0] for u in unigrams])
print("-> loses order, just words\n")

print("Bigrams:", bigrams)
print("-> keeps word pairs, context")
```

```python
# Example 6: Real example

from nltk import ngrams
from nltk.tokenize import word_tokenize
from collections import Counter

text = "Python is great Python is powerful Python is fun"
tokens = word_tokenize(text)

bigrams = list(ngrams(tokens, 2))
counts = Counter(bigrams)

print("Text:", text)
print("\nMost common bigrams:")
for bigram, count in counts.most_common():
    print(f"  {bigram}: {count} times")
```

```python
# Example 7: Bigrams vs Trigrams

from nltk import ngrams
from nltk.tokenize import word_tokenize

text = "machine learning is fun"
tokens = word_tokenize(text)

bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

print("Text:     ", text)
print("Tokens:   ", tokens)
print("\nBigrams (2-grams):")
print(bigrams)
print("\nTrigrams (3-grams):")
print(trigrams)
```

```python
# Summary
# 1. N-gram: sequence of N words
# 2. Unigrams (1-gram): single words
# 3. Bigrams (2-gram): pairs of words
# 4. Trigrams (3-gram): triplets of words
# 5. More context = more specific patterns
# 6. Useful for: autocomplete, next word prediction, language models
```
