```python
# Lemmatization in NLP

# Lemmatization reduces a word to its dictionary base form (its "lemma").
# Unlike stemming, it uses vocabulary + grammar rules, so the result is always a real word.

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["running", "ran", "runs", "better", "geese"]

print("Word      -> Lemma")
for w in words:
    print(f"{w:<10} -> {lemmatizer.lemmatize(w)}")
```

```python
# Example 1: Lemmatization needs the part of speech (POS) to work well

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

word = "better"

print("Default (treated as noun):", lemmatizer.lemmatize(word))
print("As an adjective (pos='a'):", lemmatizer.lemmatize(word, pos="a"))

# Without the right POS, "better" looks already lemmatized (it's a valid noun/adverb too)
```

```python
# Example 2: Verbs shrink a lot when you tell the lemmatizer the POS

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

verbs = ["running", "ran", "runs", "studies", "was"]

print("Word      -> Lemma (as verb)")
for v in verbs:
    print(f"{v:<10} -> {lemmatizer.lemmatize(v, pos='v')}")
```

```python
# Example 3: Lemmatizing a full sentence

from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

text = "The striped bats are hanging on their feet for best"
tokens = word_tokenize(text)

lemmas = [lemmatizer.lemmatize(t, pos="v") for t in tokens]

print("Original:", ' '.join(tokens))
print("Lemmas:  ", ' '.join(lemmas))
```

```python
# Example 4: Lemmatization vs stemming

from nltk.stem import WordNetLemmatizer, PorterStemmer

lemmatizer = WordNetLemmatizer()
stemmer = PorterStemmer()

words = ["studies", "studying", "caring", "flies", "geese"]

print(f"{'Word':<10}{'Stem':<10}{'Lemma'}")
for w in words:
    print(f"{w:<10}{stemmer.stem(w):<10}{lemmatizer.lemmatize(w, pos='v')}")

# Stemming chops off endings with rules and can produce non-words ("studi").
# Lemmatization looks up the real dictionary form ("study").
```

```python
# Example 5: Auto-detect POS instead of guessing it by hand

import nltk
from nltk import pos_tag, word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN  # default

def lemmatize_sentence(text):
    tokens = word_tokenize(text)
    tagged = pos_tag(tokens)
    return [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in tagged]

text = "The children were playing and studies were better than expected"
print("Original:", text)
print("Lemmas:  ", ' '.join(lemmatize_sentence(text)))
```

```python
# Summary
# 1. Lemmatization reduces words to their real dictionary base form (the "lemma")
# 2. It needs the part of speech (noun, verb, adjective...) to work correctly
# 3. Unlike stemming, the output is always a valid word
# 4. Use it when you need clean, human-readable base forms (search, topic modeling, chatbots)
```
