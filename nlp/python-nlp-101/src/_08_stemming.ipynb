```python
# Stemming in NLP

# Stemming reduces a word to its root form by chopping off prefixes/suffixes
# using fixed rules. It's fast but can produce forms that aren't real words.

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["running", "runs", "ran", "easily", "studies"]

print("Word      -> Stem")
for w in words:
    print(f"{w:<10} -> {stemmer.stem(w)}")
```

```python
# Example 1: Stemming a full sentence

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

text = "The striped bats are hanging on their feet for best"
tokens = word_tokenize(text)

stems = [stemmer.stem(t) for t in tokens]

print("Original:", ' '.join(tokens))
print("Stems:   ", ' '.join(stems))
```

```python
# Example 2: Different stemmers give different results

from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer

porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer("english")

words = ["studies", "studying", "caring", "flies", "generously"]

print(f"{'Word':<12}{'Porter':<12}{'Lancaster':<12}{'Snowball'}")
for w in words:
    print(f"{w:<12}{porter.stem(w):<12}{lancaster.stem(w):<12}{snowball.stem(w)}")

# Porter: oldest, mildest rules
# Lancaster: most aggressive, shortest stems
# Snowball ("Porter2"): improved, more consistent version of Porter
```

```python
# Example 3: Stemming can produce non-words

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["studies", "caring", "generously", "universal", "university"]

print("Word         -> Stem")
for w in words:
    print(f"{w:<12} -> {stemmer.stem(w)}")

# "studies" -> "studi" and "caring" -> "care" both look plausible,
# but "studi" isn't a real English word. Stemming trades accuracy for speed.
```

```python
# Example 4: Stemming vs lemmatization

from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["studies", "studying", "caring", "flies", "geese"]

print(f"{'Word':<10}{'Stem':<10}{'Lemma'}")
for w in words:
    print(f"{w:<10}{stemmer.stem(w):<10}{lemmatizer.lemmatize(w, pos='v')}")

# Stemming is fast, rule-based, and can produce non-words ("studi").
# Lemmatization is slower but always returns a real dictionary word.
```

```python
# Example 5: Create a reusable stemming function

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

def stem_text(text):
    tokens = word_tokenize(text)
    return ' '.join(stemmer.stem(t) for t in tokens)

texts = [
    "the dogs are running",
    "I am studying hard",
    "connections and connected"
]

for t in texts:
    print(f"{t:<30} -> {stem_text(t)}")
```

```python
# Summary
# 1. Stemming chops words down to a root form using fixed rules
# 2. It's fast and simple, but the output may not be a real word
# 3. Different stemmers (Porter, Lancaster, Snowball) give different results
# 4. Use it when speed matters more than readability (search indexing, IR systems)
# 5. Prefer lemmatization when you need clean, human-readable base forms
```
