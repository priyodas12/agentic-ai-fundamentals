```python
import numpy as np
```

```python
# Same symbols, different meaning: lists join/repeat, arrays do math.

a = [1, 2, 3]
b = [4, 5, 6]
print(a + b)  # [1, 2, 3, 4, 5, 6]   -> joined
print(a * 2)  # [1, 2, 3, 1, 2, 3]   -> repeated

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print(type(x))
print(x + y)  # [5 7 9]              -> added
print(x * 2)  # [2 4 6]              -> scaled
```

```python
# List slice gives a copy; array slice gives a live window into the same memory.
lst = [1, 2, 3, 4]
s = lst[1:3]
s[0] = 99
print(lst)  # [1, 2, 3, 4]  -> original untouched

arr = np.array([1, 2, 3, 4])
v = arr[1:3]
v[0] = 99
print(arr)  # [ 1 99  3  4] -> original changed
```

```python
# Lists hold anything; arrays force one type and silently truncate what does not fit.
lst = [1, "two", 3.0, None]  # perfectly legal

arr = np.array([1, 2, 3])
print(arr.dtype)  # int64
arr[0] = 3.9
print(arr)  # [3 2 3] -> truncated to int, no error
```

```python
# Nested lists need one lookup per level; arrays know their own shape.
nested = [[1, 2], [3, 4]]
print(nested[1][0])  # 3  -> two lookups

m = np.array([[1, 2], [3, 4]])
print(m[1, 0])  # 3  -> one lookup
print(m.shape)  # (2, 2)
print(m.T)  # transpose, free
```
