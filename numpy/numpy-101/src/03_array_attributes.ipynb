```python
# import numpy library
import numpy as np
```

```python
# ndim: number of dimensions (axes) of an array

print(np.ones((3, 3, 3)).ndim)
```

```python
# shape: size of the array along each dimension
print(np.ones((5, 3)).shape)
```

```python
# size: total number of elements in the array (product of all dimensions)
print(np.ones((3, 3)).size)
```

```python
from numpy import float64

# itemsize: size in bytes of each element in the array
print(np.random.random((3, 3, 4, 5)).itemsize)
print(np.arange(1, 39, dtype=float64).itemsize)
```
