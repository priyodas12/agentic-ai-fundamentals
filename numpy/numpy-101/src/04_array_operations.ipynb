```python
# set up two int32 arrays for arithmetic operations below
import numpy as np
from numpy import int32

a1 = np.arange(12, dtype=int32).reshape(3, 4)  # 0..11 as a 3x4 array
b1 = np.arange(12, 24, dtype=int32).reshape(3, 4)  # 12..23 as a 3x4 array

print(a1)
print(b1)
```

```python
# scalar operation: multiplying an array by a single number applies it to every element (broadcasting)
a1 * 23
```

```python
# comparison operator: elementwise equality check, returns boolean array
a1 == 9
```

```python
# comparison operator: elementwise greater-than check, returns boolean array
a1 > 2
```

```python
# vector operation: elementwise arithmetic between two same-shaped arrays (no loops needed)

print(f"sum: {a1 + b1}")  # elementwise addition
print(f"subtract: {a1 - b1}")  # elementwise subtraction
print(f"multiply: {a1 * b1}")  # elementwise (Hadamard) multiplication, not matrix multiplication
```

```python
# max: largest element in the array (aggregate reductions like min/max/sum/prod collapse an array to fewer values)

np.max(a1)
```

```python
# min: smallest element in the array
np.min(b1)
```

```python
# sum: total of all elements in the array
np.sum(a1)
```

```python
# prod: product of all elements in the array
np.prod(b1)
```

```python
# max along an axis: axis=1 gives row-wise max, axis=0 gives column-wise max
np.max(a1, axis=1)
np.max(b1, axis=0)
```

```python
# mean: average of all elements in the array
np.mean(b1)
```

```python
# median: middle value of the sorted elements in the array
np.median(b1)
```

```python
# dot product: for two vectors, returns a scalar (sum of elementwise products),
# measures alignment

v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

dot_product = np.dot(v1, v2)

dot_product
```

```python
# cross product: for two 3D vectors, returns a vector perpendicular to both (used in physics/graphics for normals, torque, angular velocity)

v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

cross_product = np.cross(v1, v2)

cross_product
```

```python
# log: natural logarithm (base e) of each element
np.log(b1)
```

```python
# exp: e raised to the power of each element (inverse of log)
np.exp(b1)
```

```python
# ceil and floor: round each element up (ceil) or down (floor) to the nearest integer

c1 = np.array([1.2, 1.8, -1.2, -1.8, 2.5])

print(f"ceil: {np.ceil(c1)}")  # rounds up towards positive infinity
print(f"floor: {np.floor(c1)}")  # rounds down towards negative infinity
```
