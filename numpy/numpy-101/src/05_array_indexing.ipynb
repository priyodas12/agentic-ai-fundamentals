```python
# import numpy library
import numpy as np
```

```python
# 1D array: a simple sequence of values
a1 = np.arange(10)
a1
```

```python
# 2D array: rows and columns, indexed as [row, column]
b1 = np.arange(15, 30).reshape(3, 5)
b1
```

```python
# 3D array: a stack of 2D arrays, indexed as [depth, row, column]
c1 = np.arange(8).reshape(2, 2, 2)
c1
```

```python
# slicing a 1D array: elements from index 1 up to (not including) index 4
a1[1:4]
```

```python
# view the full 2D array again before indexing into it
b1
```

```python
# indexing a 2D array: single element at row 2, column 2
b1[2, 2]
```

```python
# slicing a 2D array: all columns (:) of row 0
b1[0, :]
```

```python
# slicing a 2D array: all rows (:) of column 0
b1[:, 0]
```
