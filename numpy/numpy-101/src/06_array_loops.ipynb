```python
# import numpy library
import numpy as np
```

```python
# 1D array: a simple sequence of values
a1 = np.arange(12)
a1
```

```python
# looping a 1D array: yields each scalar element in order
for num in a1:
    print(num)
```

```python
# looping a 2D array: yields each row (a 1D array), not individual elements
b1 = a1.reshape(3, 4)

for num in b1:
    print(num)
```

```python
# 3D array: a stack of 2D arrays
c1 = np.arange(60).reshape(3, 4, 5)
c1
```

```python
# looping a 3D array: yields each 2D "slice" along the first axis, not individual elements
for num in c1:
    print(num)
```

```python
# np.nditer: flattens iteration over any n-dimensional array to visit every scalar element
for num in np.nditer(c1):
    print(num)
```

```python
# view the 2D array again before transposing it
b1
```

```python
# transpose: flips rows and columns, converting a 3x4 array into a 4x3 array
np.transpose(b1)
```

```python
# ravel: flattens an n-dimensional array into a single 1D array
np.ravel(c1)
```

```python
# reshape: functional form of .reshape(), returns a new view with the given shape
np.reshape(a1, (3, 4))
```
