```python
# import numpy library
import numpy as np
```

```python
# two simple 1D arrays to stack together
a1 = np.array([1, 2, 3])
a2 = np.array([4, 5, 6])

print(a1)
print(a2)
```

```python
# vstack: stack arrays on top of each other (row-wise), adds a new row
np.vstack((a1, a2))
```

```python
# hstack: stack arrays side by side (column-wise), joins them into one longer row
np.hstack((a1, a2))
```

```python
# hstack on 1D arrays only concatenates them, so lengths can differ
np.hstack((np.array([1, 2, 3]), np.array([4, 5])))
```

```python
# vstack on 1D arrays needs equal length, since each becomes a row and rows must match in width
np.vstack((np.array([1, 2, 3]), np.array([4, 5])))
```

```python
# two 2D arrays with the same number of columns (3) but different number of rows
m1 = np.array([[1, 2, 3], [4, 5, 6]])  # 2x3
m2 = np.array([[7, 8, 9]])  # 1x3

print(m1)
print(m2)
```

```python
# vstack on 2D arrays: number of columns must match, rows can differ -> result is 3x3
np.vstack((m1, m2))
```

```python
# two 2D arrays with the same number of rows (2) but different number of columns
n1 = np.array([[1, 2], [3, 4]])  # 2x2
n2 = np.array([[5, 6, 7], [8, 9, 10]])  # 2x3

print(n1)
print(n2)
```

```python
# hstack on 2D arrays: number of rows must match, columns can differ -> result is 2x5
np.hstack((n1, n2))
```
