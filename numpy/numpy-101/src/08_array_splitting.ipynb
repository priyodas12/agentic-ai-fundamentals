```python
# import numpy library
import numpy as np
```

```python
# 1D array to split into equal parts
a1 = np.arange(9)
a1
```

```python
# split: divide a 1D array into n equal-sized parts (must divide evenly)
np.split(a1, 3)
```

```python
# split at specific index positions instead of equal parts
np.split(a1, [2, 5])
```

```python
# 2D array to split both horizontally and vertically
b1 = np.arange(12).reshape(3, 4)
b1
```

```python
# hsplit: split a 2D array column-wise (left to right) into equal parts
np.hsplit(b1, 2)
```

```python
# vsplit: split a 2D array row-wise (top to bottom) into equal parts
np.vsplit(b1, 3)
```

```python
# splitting into unequal parts must divide evenly for np.split, otherwise it raises an error
np.split(a1, 4)
```
