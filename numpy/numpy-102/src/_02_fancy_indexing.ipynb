# Fancy Indexing

Fancy indexing lets you index a NumPy array using another array (or list) of indices, instead of a single integer or a slice. It always returns a **copy** of the data, not a view.

```python
# NumPy import used for all fancy indexing examples
import numpy as np
```

## Fancy indexing on 1-D arrays

```python
# Create a simple 1-D array to index into
arr = np.arange(10, 20)
print(arr)

# Select elements at indices 1, 3, and 5 using a list of indices
indices = [1, 3, 5]
print(arr[indices])
```

```python
# Negative indices work too, and the order of indices controls the order of the result
print(arr[[-1, -2, 0]])
```

## Fancy indexing on 2-D arrays

```python
# Create a 4x4 matrix to index into
matrix = np.arange(1, 17).reshape(4, 4)
print(matrix)

# Select rows 0, 2, and 3 (in that order)
print(matrix[[0, 2, 3]])
```

```python
# Pairing row indices with column indices selects individual elements:
# (0, 1), (1, 2), (2, 3) -> element-wise pairing, not a cross product
rows = [0, 1, 2]
cols = [1, 2, 3]
print(matrix[rows, cols])
```

```python
# Combining fancy indexing with a slice: pick specific rows, all columns
print(matrix[[0, 3], :])
```

## Fancy indexing with boolean masks

```python
# Create a 1-D array to filter with a boolean mask
arr2 = np.arange(1, 11)

# Boolean mask selects only the elements where the condition is True
mask = arr2 % 2 == 0
print(mask)
print(arr2[mask])
```

## Modifying arrays with fancy indexing

```python
# Create a 1-D array to modify via fancy indexing
arr3 = np.arange(10)

# Assign new values at selected indices in a single vectorized operation
arr3[[0, 2, 4]] = 0
print(arr3)
```

## Fancy indexing returns a copy, not a view

```python
# Create an array and select a subset of it via fancy indexing
original = np.arange(5)
fancy_result = original[[0, 1, 2]]

# Modifying the fancy-indexed result does NOT affect the original array
fancy_result[0] = 100
print("original:", original)
print("fancy_result:", fancy_result)
```
