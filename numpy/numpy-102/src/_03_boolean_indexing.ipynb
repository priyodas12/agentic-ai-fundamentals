```python
# NumPy import used for boolean indexing examples
import numpy as np
```

```python
# randint(start,end,how_many_numbers_you_want)
# Generate 24 random integers between 1 and 99, arranged as a 6x4 matrix
a = np.random.randint(1, 100, 24).reshape(6, 4)

a
```

```python
# Element-wise comparison builds a boolean mask, same shape as `a`
a > 39
```

```python
# Using the boolean mask to index `a` flattens the result to only the elements > 50
gt_50 = a[a > 50]

gt_50
```

```python
# Chain another boolean mask on the filtered result to keep only even numbers
gt_50[gt_50 % 2 == 0]
```

```python
# Combine multiple conditions with & to select elements > 50 that are also divisible by 7
a[(a > 50) & (a % 7 == 0)]
```

```python
# Use ~ to negate a condition: select elements that are NOT divisible by 7
a[~(a % 7 == 0)]
```
