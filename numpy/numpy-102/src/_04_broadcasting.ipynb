```python
# NumPy import used for broadcasting examples
import numpy as np
```

```python
# Create a 3x4 array
a = np.arange(12).reshape(3, 4)
a
```

```python
# Create another 3x4 array, same shape as `a`
b = np.arange(12, 24).reshape(3, 4)
b
```

```python
# Broadcasting Rule 1: if shapes are already identical, element-wise ops just work directly
a + b
```

```python
# Create a 4x4 array (deliberately mismatched with `a`'s 3x4 shape) to trigger a broadcasting failure
c = np.arange(16).reshape(4, 4)
c
```

```python
# This fails Broadcasting Rule 2: comparing shapes from the trailing dimension,
# (3, 4) vs (4, 4) -> the leading dims (3 vs 4) are neither equal nor equal to 1,
# so the shapes are incompatible and NumPy raises a ValueError instead of broadcasting
a + c
```

## Fixing it: valid broadcasting examples

Broadcasting rules, in simple terms (NumPy compares shapes starting from the right side):
1. If one array has fewer dimensions, NumPy pretends it has extra 1s added to the left of its shape.
2. Going dimension by dimension, two sizes can be used together if they are the same number, or if one of them is 1 (a size of 1 just gets repeated/stretched to match the other).
3. If any pair of sizes is different and neither one is 1, NumPy gives up and throws an error (that's exactly what happened above with `a + c`).

```python
# Rule 1 in action: a scalar has 0 dimensions, so it's treated as shape (1,) then
# padded/stretched to match every dimension of `a` -> (3, 4)
a + 10
```

```python
# Rule 1 + Rule 2: `row` has shape (4,) -> padded on the left to (1, 4),
# then the size-1 dimension is stretched to match `a`'s 3 rows
row = np.arange(1, 5)
a + row
```

```python
# Rule 2 with a column vector: `col` has shape (3, 1), `a` has shape (3, 4) ->
# the trailing dims (1 vs 4) are compatible because one of them is 1, so `col`
# is stretched across all 4 columns
col = np.array([[0], [10], [20]])
a + col
```

```python
# Combining Rule 1 and Rule 2 together:
# `p` has shape (2, 3, 4), `q` has shape (1, 4)
# Rule 1: `q` has fewer dims, so it's padded on the left -> (1, 1, 4)
# Rule 2: comparing (2, 3, 4) vs (1, 1, 4) from the right -> 4==4 (match),
#         3 vs 1 (stretch), 2 vs 1 (stretch) -> `q` is stretched across both
#         the first and second dimensions of `p`
p = np.arange(24).reshape(2, 3, 4)
q = np.array([[1, 2, 3, 4]])
p + q
```
