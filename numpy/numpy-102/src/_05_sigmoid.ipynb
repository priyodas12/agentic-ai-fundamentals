# Sigmoid Function

The sigmoid function squashes any real number into the range `(0, 1)`:

$$\sigma(x) = \frac{1}{1 + e^{-x}}$$

**Shape:**
- Large negative `x` -> output close to `0`
- `x = 0` -> output exactly `0.5`
- Large positive `x` -> output close to `1`
- It's an S-shaped (logistic) curve, smooth and differentiable everywhere.

**Common uses:**
- **Neural networks (activation function):** introduces non-linearity so a network can learn complex patterns, not just linear combinations of inputs.
- **Binary classification output layer:** since the output is between 0 and 1, it's naturally interpreted as a probability (e.g. "probability this email is spam").
- **Logistic regression:** the sigmoid is literally what turns a linear model's raw score into a probability.
- **Gating in RNNs/LSTMs:** used to decide "how much" of a signal to let through (values near 0 = block, near 1 = pass).

Using NumPy here lets `sigmoid()` operate on a whole array at once (vectorized), instead of looping element by element in plain Python.

```python
# NumPy import used for the vectorized exp() in the sigmoid implementation
import numpy as np
```

```python
# sigmoid(x) = 1 / (1 + e^-x), applied element-wise via NumPy's vectorized exp()
def sigmoid(array):
    return 1 / (1 + np.exp(-(array)))


# Try it on 0..9: notice values approach 1 as x grows (never quite reaching it)
a = np.arange(10)

sigmoid(a)
```
