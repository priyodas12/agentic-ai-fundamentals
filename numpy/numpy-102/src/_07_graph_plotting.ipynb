```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)
y = x
```

```python
plt.plot(x, y)
```

```python
m = np.linspace(-10, 10, 100)

n = m + 100

plt.plot(m, n)
```

```python
n = m**2

plt.plot(m, n)
```

```python
y = np.sin(x)
plt.plot(x, y)
```
