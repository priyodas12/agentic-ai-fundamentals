```python
# NumPy is used for fast, vectorized array math
import numpy as np
```

```python
# Create a 1D array of 15 random integers between 1 and 99
a = np.random.randint(1, 100, 15)

print(a)

# np.sort returns a new sorted array in ascending order (original array is left unchanged)
sorted_a = np.sort(a)

sorted_a
```

```python
# Create 15 random integers and reshape them into a 3x5 2D array

b = np.random.randint(1, 100, 15).reshape(3, 5)
b
```

```python
# Descending sort: np.sort has no "reverse" flag, so sort ascending then reverse with slicing

np.sort(a)[::-1]
```

```python
# axis: row wise sorting 1

np.sort(b,axis=1,kind='mergesort')
```

```python
c = np.random.randint(1,100,24).reshape(4,6)

c
```

```python
# Append a new row to c along axis=0 — the new row must have the same number of columns (6)
np.append(c, np.random.randint(1, 100, (4, 6)), axis=0)
```
