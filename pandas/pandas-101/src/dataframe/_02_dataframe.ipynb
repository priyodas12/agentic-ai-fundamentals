## `loc` vs `iloc`

- **`iloc`** — selects by **integer position**, like list indexing. `movies.iloc[101]` gets the row at position 101, regardless of what its index label is. Slicing with `iloc` **excludes** the stop (e.g. `movies.iloc[0:5]` gives 5 rows).
- **`loc`** — selects by **label**, i.e. the actual index/column value. `movies.loc['Abhay']` looks up the row whose index label is `'Abhay'`. Slicing with `loc` **includes** the stop (e.g. `movies.loc['Abhay':'Ajnabee']` includes both endpoints).

Rule of thumb: use `iloc` when you care about row/column position, use `loc` when you care about the actual labels (names, dates, IDs, etc.).

```python
# select a row by label using loc (movie_name set as index just for this lookup)
movies.set_index('movie_name').loc['Abhay']
```
