```python
# import pandas and numpy libraries
import pandas as pd
import numpy as np
```

```python
# create a series from a list with custom integer index
crypto_currency = ['btc','eth','solana','dogecoin','memecoin','usdt','usdc']

crypto_currency_series = pd.Series(crypto_currency,index=[101,102,103,302,304,403,405])

crypto_currency_series
```

```python
# create a named series with custom labels as index
marks = [55,65,87,39]
subjects = ['bengali','english','spanish','german']


marksheet = pd.Series(marks, subjects,name='marksheet')
```

```python
# create a series from a dictionary, keys become the index
metadata={'name':'priyo','age':100,'stream':'engineering'}

a = pd.Series(metadata)

a
```
