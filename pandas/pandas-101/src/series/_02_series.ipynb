```python
import pandas as pd

csv_path = "../data/monthly_sales.csv"

# month becomes the index, sales becomes the values
s = pd.read_csv(csv_path, index_col="month")["sales"]

print(s)
print(type(s))  # <class 'pandas.core.series.Series'>
print(s["Mar"])  # 980
print(s.mean())  # average sales
```

```python
# Old pandas (<2.0) used to support: pd.read_csv(csv_path, index_col="month", squeeze=True)
# squeeze was removed in pandas 2.0 - use .squeeze("columns") instead
s2 = pd.read_csv(csv_path, index_col="month").squeeze("columns")

print(s2)
print(type(s2))  # <class 'pandas.core.series.Series'>
```

```python
# preview the first 5 values
s2.head()
```

```python
# preview the last 5 values
s2.tail()
```

```python
# get a random sample of 3 values
s2.sample(3)
```

```python
# count occurrences of each unique value
s2.value_counts()
```

```python
# load a column as a series indexed by movie_name
movies = pd.read_csv('../data/bollywood_movies.csv', index_col="movie_name")["lead_actor"]
print(movies.head())
print(type(movies))  # <class 'pandas.core.series.Series'>
```

```python
# get a random sample of 4 movies
movies.sample(4)
```

```python
# sort the series by its index
movies.sort_index()
```

```python
# count occurrences of each lead actor
movies.value_counts()
```

```python
# count non-null values in the series
movies.count()
```
