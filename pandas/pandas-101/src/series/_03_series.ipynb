```python
# series indexing
import numpy as np
import pandas as pd

```

```python
# create a series of 12 random integers
x = pd.Series(np.random.randint(1,100,12),name='x')

x
```

```python
# indexing

x[0]
```

```python
# slicing

x[2:7]
```

```python
# negative slicing, last 5 values
x[-5:]
```

```python
# fancy indexing

x[::3]
```

```python
# fancy indexing with a list of positions
x[[1,4,6,9]]
```
