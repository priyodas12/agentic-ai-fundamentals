# write an example for how variable can be termed (naming rules)

- Variable names can contain letters, digits, and underscores, but cannot start with a digit.
- Names are case-sensitive (`age` and `Age` are different variables).
- Cannot use Python reserved keywords (e.g. `class`, `for`, `if`) as variable names.

## Single and multiline commands

- A single statement usually sits on one line, ending at the newline.
- Multiple statements can share one line using a semicolon `;`.
- A single statement can span multiple lines using `\` (line continuation) or by wrapping in `()`, `[]`, `{}` (implicit continuation, preferred).

```python
# single statement
x = 10

# multiple statements on one line, separated by ;
a = 1
b = 2
c = 3
print(a, b, c)

# multiline using backslash continuation
total = 1 + 2 + 3 + 4 + 5 + 6
print(total)

# multiline using implicit continuation (parentheses) - preferred
total = 1 + 2 + 3 + 4 + 5 + 6
print(total)

numbers = [
    1,
    2,
    3,
    4,
    5,
    6,
]
print(numbers)
```

## Python indentation

- Python uses indentation (whitespace) instead of `{}` to define blocks of code.
- Standard convention is 4 spaces per indentation level.
- All statements within the same block must be indented at the same level, or Python raises an `IndentationError`.

```python
# correct indentation
age = 20
if age >= 18:
    print("You are an adult")
    print("You can vote")
else:
    print("You are a minor")

# nested indentation
for i in range(3):
    if i % 2 == 0:
        print(i, "is even")
    else:
        print(i, "is odd")

# inconsistent indentation raises IndentationError (uncomment to test)
# if age >= 18:
#     print("adult")
#      print("mismatched indent")  # IndentationError
```

## Basic syntax rules

- Python is case-sensitive.
- Comments start with `#` and go to the end of the line.
- Colons `:` are required to start a new block (`if`, `for`, `while`, `def`, `class`, etc.).
- Statement blocks are defined by indentation, not braces.
- Whitespace inside a line is generally not significant (except leading indentation).

```python
# comment example - this whole line is ignored by Python
x = 5  # inline comment: assigns 5 to x


# colon required to start a block
def greet(name):
    return f"Hello, {name}!"


print(greet("World"))

# whitespace within a line doesn't matter
y = 1 + 2
z = 1 + 2
print(y == z)
```

## Understanding semantics for Python

Syntax is about *form* (is the code structured correctly?). Semantics is about *meaning* (what does the code actually do when it runs?). Code can be syntactically valid but semantically wrong (it runs, but produces the wrong result or an unexpected error).

```python
# syntactically valid, but semantically wrong: dividing by the wrong variable
price = 100
quantity = 0
# this line is valid Python syntax, but raises a runtime error (semantic issue)
try:
    average = price / quantity
except ZeroDivisionError as e:
    print("Semantic/runtime error:", e)

# syntactically valid, but wrong meaning: using '=' vs '==' correctly matters
a = 5  # assignment (semantics: store 5 in a)
print(a == 5)  # comparison (semantics: check equality) -> True

# dynamic typing: same variable name, different meaning/type over time
value = 10  # value is an int here
print(type(value))
value = "ten"  # now value is a str - Python allows this, syntax is fine
print(type(value))
```

## Common syntax errors

Below are frequent mistakes that raise `SyntaxError` / `IndentationError`. Each broken snippet is run through `exec()` inside a `try/except` so we can safely capture and print the error message without stopping the notebook.

```python
def show_error(label, code):
    try:
        exec(code)
    except (SyntaxError, IndentationError) as e:
        print(f"{label}: {type(e).__name__}: {e}")


# 1. missing colon
show_error("Missing colon", "if True\n    print('hi')")

# 2. mismatched indentation
show_error("Bad indentation", "if True:\n    print('a')\n  print('b')")

# 3. unclosed parenthesis
show_error("Unclosed parenthesis", "print('hello'")

# 4. using a reserved keyword as a variable name
show_error("Reserved keyword as name", "class = 5")

# 5. missing quotes around a string
show_error("Unterminated string", "name = 'Priyo")

# 6. invalid assignment target
show_error("Invalid assignment target", "5 = x")
```

## `compile()` vs `exec()`

- `exec()` parses, compiles, **and runs** a string of code in one step.
- `compile()` only parses and compiles a string into a code object — it does **not** execute anything. This makes it useful purely for syntax checking, or for compiling once and running many times with `exec()`.

```python
# compile() checks syntax without running the code
try:
    code_obj = compile("if True\n    print('hi')", "<string>", "exec")
    print("Compiled successfully:", code_obj)
except SyntaxError as e:
    print("SyntaxError caught by compile():", e)
    # -> SyntaxError caught by compile(): expected ':' (<string>, line 1)

# valid code: compile() just builds a code object, nothing runs yet
code_obj = compile("print('this has not run yet')", "<string>", "exec")
print("compile() returned:", code_obj)
# -> compile() returned: <code object <module> at 0x..., file "<string>", line 1>
print(
    "Notice nothing printed above from the compiled code - compile() does not execute it"
)
# -> Notice nothing printed above from the compiled code - compile() does not execute it

# exec() can now run that pre-compiled code object
exec(code_obj)
# -> this has not run yet

# compiling once, executing many times (a practical use case)
loop_code = compile("total += i", "<string>", "exec")
total = 0
for i in range(5):
    exec(loop_code)
print("total after running compiled code 5 times:", total)
# -> total after running compiled code 5 times: 10
```

```python
# Python infers the type from the assigned value - no declaration needed
a = 10  # int, inferred from the integer literal
b = 3.14  # float, inferred from the decimal literal
c = "hello"  # str, inferred from the quotes
d = True  # bool
e = [1, 2, 3]  # list
f = {"x": 1}  # dict

for name, val in [("a", a), ("b", b), ("c", c), ("d", d), ("e", e), ("f", f)]:
    print(f"{name} = {val!r:<15} inferred type: {type(val).__name__}")
    # -> a = 10              inferred type: int
    # -> b = 3.14            inferred type: float
    # -> c = 'hello'         inferred type: str
    # -> d = True            inferred type: bool
    # -> e = [1, 2, 3]       inferred type: list
    # -> f = {'x': 1}        inferred type: dict

# type inference is dynamic - rebinding a name can change its inferred type
x = 5
print(x, type(x).__name__)  # -> 5 int
x = "now a string"
print(x, type(x).__name__)  # -> now a string str

# inference also applies to expressions, not just literals
result = 10 / 3  # division always infers float
print(result, type(result).__name__)  # -> 3.3333333333333335 float
result = 10 // 3  # floor division infers int (when operands are int)
print(result, type(result).__name__)  # -> 3 int


# duck typing: Python cares about behavior, not declared type
def describe(value):
    return f"{value!r} is of type {type(value).__name__}"


for item in [42, 3.14, "text", [1, 2], (1, 2), {1, 2}, None]:
    print(describe(item))
    # -> 42 is of type int
    # -> 3.14 is of type float
    # -> 'text' is of type str
    # -> [1, 2] is of type list
    # -> (1, 2) is of type tuple
    # -> {1, 2} is of type set
    # -> None is of type NoneType


# optional type hints: documentation/tooling only, NOT enforced at runtime
def add(x: int, y: int) -> int:
    return x + y


print(add(2, 3))  # -> 5 (works)
print(
    add("2", "3")
)  # -> 23 (string concatenation, not addition - hints are not checked by Python itself)
```
