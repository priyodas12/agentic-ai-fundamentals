# Understanding Variable Types

Python is **dynamically typed** — you don't declare a variable's type; it's inferred from the value assigned. Below we explore the built-in types with examples.

## 1. Basic variable assignment

```python
name = "Priyo"  # str
age = 30  # int
height = 5.9  # float
is_developer = True  # bool

print(name, age, height, is_developer)  # Priyo 30 5.9 True
print(
    type(name), type(age), type(height), type(is_developer)
)  # <class 'str'> <class 'int'> <class 'float'> <class 'bool'>
```

## 2. Numeric types: int, float, complex

```python
whole_number = 42
decimal_number = 3.14159
complex_number = 2 + 3j

print(whole_number, type(whole_number))  # 42 <class 'int'>
print(decimal_number, type(decimal_number))  # 3.14159 <class 'float'>
print(complex_number, type(complex_number))  # (2+3j) <class 'complex'>

# arithmetic between numeric types
print(whole_number + decimal_number)  # int + float -> float          # 45.14159
print(whole_number / 4)  # true division -> float       # 10.5
print(whole_number // 4)  # floor division -> int        # 10
print(whole_number % 4)  # modulus                      # 2
```

## 3. Strings (str)

```python
greeting = "Hello"
target = "World"
combined = f"{greeting}, {target}!"  # f-string formatting

print(combined)  # Hello, World!
print(combined.upper())  # HELLO, WORLD!
print(combined.lower())  # hello, world!
print(len(combined))  # 13
print(combined[0:5])  # slicing        # Hello
print(combined.split(", "))  # ['Hello', 'World!']
```

## 4. Boolean (bool)

```python
is_raining = False
is_sunny = True

print(is_raining, type(is_raining))  # False <class 'bool'>
print(is_raining or is_sunny)  # True
print(is_raining and is_sunny)  # False
print(not is_raining)  # True

# bool is a subclass of int
print(True + True)  # 2
print(int(True))  # 1
```

## 5. Collections: list, tuple, set, dict

```python
# list: ordered, mutable
fruits = ["apple", "banana", "cherry"]
fruits.append("date")
print(fruits, type(fruits))  # ['apple', 'banana', 'cherry', 'date'] <class 'list'>

# tuple: ordered, immutable
coordinates = (10, 20)
print(coordinates, type(coordinates))  # (10, 20) <class 'tuple'>

# set: unordered, unique elements
unique_numbers = {1, 2, 2, 3, 3, 3}
print(unique_numbers, type(unique_numbers))  # {1, 2, 3} <class 'set'>

# dict: key-value pairs
person = {"name": "Alice", "age": 30, "city": "Metropolis"}
print(
    person, type(person)
)  # {'name': 'Alice', 'age': 30, 'city': 'Metropolis'} <class 'dict'>
print(person["name"])  # Alice
```

## 6. NoneType

```python
result = None
print(result, type(result))  # None <class 'NoneType'>
print(result is None)  # True
```

## 7. Dynamic typing

Since Python is dynamically typed, a variable can be reassigned to a different type at any point in the program — the variable name is just a label pointing to an object.

```python
value = 100
print(value, type(value))  # 100 <class 'int'>

value = "now I'm a string"
print(value, type(value))  # now I'm a string <class 'str'>
```

## 8. Type conversion (casting)

Python supports two kinds of type conversion:

- **Implicit conversion** — Python automatically converts one type to another when it's "safe" to do so (no data loss), e.g. `int` → `float` during arithmetic.
- **Explicit conversion** — you manually convert using built-in functions: `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`, `dict()`.

### 8.1 Implicit conversion

```python
count = 10  # int
price = 2.5  # float

total = count * price  # int * float -> Python implicitly converts count to float
print(total, type(total))  # 25.0 <class 'float'>

# bool is implicitly treated as int (True=1, False=0)
print(5 + True)  # 6
print(5 + False)  # 5
```

### 8.2 Explicit conversion

```python
# numeric <-> string
num_str = "25"
num_int = int(num_str)
num_float = float(num_str)
print(num_int, type(num_int))  # 25 <class 'int'>
print(num_float, type(num_float))  # 25.0 <class 'float'>
print(str(num_int) + " apples")  # 25 apples

# float -> int truncates (does not round)
print(int(9.9))  # 9
print(int(-9.9))  # -9

# numeric -> bool: 0/0.0 is False, everything else is True
print(bool(0), bool(1), bool(-5), bool(0.0))  # False True True False

# string -> bool: empty string is False, non-empty is True
print(bool(""), bool("False"), bool("hello"))  # False True True

# collections
print(list("abc"))  # ['a', 'b', 'c']
print(tuple([1, 2, 3]))  # (1, 2, 3)
print(set([1, 1, 2, 3]))  # {1, 2, 3}
print(dict([("a", 1), ("b", 2)]))  # {'a': 1, 'b': 2}
```

### 8.3 Conversion errors

Not every conversion is valid — trying to convert an incompatible string raises a `ValueError`.

```python
try:
    int("not a number")
except ValueError as e:
    print(
        f"ValueError: {e}"
    )  # ValueError: invalid literal for int() with base 10: 'not a number'
```
