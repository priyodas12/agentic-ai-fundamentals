```python
### calculator example
number1 = input("insert any number:")
number2 = input("insert any number:")

sum = int(number1) + int(number2)
sub = int(number1) - int(number2)
mul = int(number1) * int(number2)
div = int(number1) / int(number2)

print(f"sum: {sum} sub: {sub} mul: {mul} div: {div}")
```
