# Exploring Python Data Types

Goal: in this notebook we just formally mention the data types available in Python, grouped by category. Later on we will create a separate notebook for each individual data type with a deeper dive.

## 1. Numeric Types: `int`, `float`, `complex`

```python
int_val = 10
float_val = 10.5
complex_val = 10 + 5j

print(type(int_val))  # <class 'int'>
print(type(float_val))  # <class 'float'>
print(type(complex_val))  # <class 'complex'>
```

## 2. Sequence Types: `str`, `list`, `tuple`, `range`

```python
str_val = "hello"
list_val = [1, 2, 3]
tuple_val = (1, 2, 3)
range_val = range(5)

print(type(str_val))  # <class 'str'>
print(type(list_val))  # <class 'list'>
print(type(tuple_val))  # <class 'tuple'>
print(type(range_val))  # <class 'range'>
```

## 3. Mapping Type: `dict`

```python
dict_val = {"a": 1, "b": 2}

print(type(dict_val))  # <class 'dict'>
```

## 4. Set Types: `set`, `frozenset`

```python
set_val = {1, 2, 3}
frozenset_val = frozenset({1, 2, 3})

print(type(set_val))  # <class 'set'>
print(type(frozenset_val))  # <class 'frozenset'>
```

## 5. Boolean Type: `bool`

```python
bool_val = True

print(type(bool_val))  # <class 'bool'>
```

## 6. Binary Types: `bytes`, `bytearray`, `memoryview`

```python
bytes_val = b"hello"
bytearray_val = bytearray(5)
memoryview_val = memoryview(bytes(5))

print(type(bytes_val))  # <class 'bytes'>
print(type(bytearray_val))  # <class 'bytearray'>
print(type(memoryview_val))  # <class 'memoryview'>
```

## 7. None Type: `NoneType`

```python
none_val = None

print(type(none_val))  # <class 'NoneType'>
```
