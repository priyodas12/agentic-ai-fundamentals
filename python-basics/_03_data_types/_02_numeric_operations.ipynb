# Numeric Types: Operations

Exploring arithmetic, comparison, and built-in math operations on `int`, `float`, and `complex`.

```python
val_int = 7
val_float = 7.0
val_complex = 7 + 0j

print(type(val_int))  # <class 'int'>
print(type(val_float))  # <class 'float'>
print(type(val_complex))  # <class 'complex'>
print(
    val_int == val_float == val_complex
)  # True -> numerically equal despite different types
```

## 1. Basic arithmetic

```python
a = 17
b = 5

print(a + b)  # 22
print(a - b)  # 12
print(a * b)  # 85
print(a / b)  # 3.4  -> true division always returns float
print(a // b)  # 3    -> floor division
print(a % b)  # 2    -> modulus (remainder)
print(a**b)  # 1419857 -> exponentiation
```

## 2. Comparison operators

```python
print(a > b)  # True
print(a < b)  # False
print(a == b)  # False
print(a != b)  # True
print(a >= 17)  # True
print(a <= 16)  # False
```

## 3. Augmented assignment

```python
x = 10
x += 5  # x = x + 5
print(x)  # 15
x -= 3
print(x)  # 12
x *= 2
print(x)  # 24
x //= 5
print(x)  # 4
```

## 4. Built-in math helpers

```python
print(abs(-7))  # 7
print(round(3.14159, 2))  # 3.14
print(round(2.5))  # 2  -> banker's rounding (rounds to even)
print(round(3.5))  # 4
print(min(4, 9, 1))  # 1
print(max(4, 9, 1))  # 9
print(pow(2, 10))  # 1024
print(divmod(17, 5))  # (3, 2) -> (floor division, modulus)
```

## 5. The `math` module

```python
import math

print(math.sqrt(16))  # 4.0
print(math.floor(3.9))  # 3
print(math.ceil(3.1))  # 4
print(math.pi)  # 3.141592653589793
print(math.factorial(5))  # 120
```

## 6. Complex numbers

```python
z = 3 + 4j

print(z.real)  # 3.0
print(z.imag)  # 4.0
print(abs(z))  # 5.0 -> magnitude: sqrt(real**2 + imag**2)
print(z.conjugate())  # (3-4j)
print(z + (1 + 1j))  # (4+5j)
```

## 7. Mixed-type arithmetic and precision

When operands differ in type, Python promotes to the "widest" type: `int` -> `float` -> `complex`. Floats are also subject to binary floating-point rounding error.

```python
print(3 + 4.0)  # 7.0  -> int + float -> float
print(3 + (1 + 2j))  # (4+2j) -> int + complex -> complex
print(0.1 + 0.2)  # 0.30000000000000004 -> classic float precision quirk
print(round(0.1 + 0.2, 2))  # 0.3
```
