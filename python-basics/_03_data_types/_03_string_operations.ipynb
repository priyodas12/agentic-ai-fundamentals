# String (`str`) Operations

Strings are immutable, ordered sequences of characters.

```python
word = "python"

# ordered -> each character has a fixed position
print(word[0])  # p
print(word[-1])  # n

# immutable -> can't reassign a character after creation
try:
    word[0] = "P"
except TypeError as e:
    print(f"TypeError: {e}")  # TypeError: 'str' object does not support item assignment
```

## 1. Creating and indexing strings

```python
s = "Hello, World!"

print(s[0])  # H
print(s[-1])  # !
print(s[7:12])  # World  -> slicing [start:stop]
print(s[::-1])  # !dlroW ,olleH  -> reversed via step -1
```

## 2. Concatenation and repetition

```python
first = "Py"
second = "thon"

print(first + second)  # Python
print(first * 3)  # PyPyPy
```

## 3. Immutability

Strings cannot be modified in place — every "change" produces a new string object.

```python
s = "cat"
try:
    s[0] = "b"
except TypeError as e:
    print(f"TypeError: {e}")  # TypeError: 'str' object does not support item assignment

s = "b" + s[1:]
print(s)  # bat
```

## 4. Case and whitespace methods

```python
raw = "  Hello World  "

print(raw.upper())  # '  HELLO WORLD  '
print(raw.lower())  # '  hello world  '
print(raw.strip())  # 'Hello World'
print(raw.lstrip())  # 'Hello World  '
print(raw.rstrip())  # '  Hello World'
print(raw.strip().title())  # 'Hello World'
```

## 5. Searching and replacing

```python
s = "the quick brown fox"

print(s.find("quick"))  # 4  -> index of first match, -1 if not found
print(s.find("slow"))  # -1
print(s.count("o"))  # 2
print(s.replace("fox", "dog"))  # the quick brown dog
print(s.startswith("the"))  # True
print(s.endswith("fox"))  # True
print("quick" in s)  # True
```

## 6. Splitting and joining

```python
csv_row = "apple,banana,cherry"

fruits = csv_row.split(",")
print(fruits)  # ['apple', 'banana', 'cherry']
print("-".join(fruits))  # apple-banana-cherry

lines = "line1\nline2\nline3"
print(lines.splitlines())  # ['line1', 'line2', 'line3']
```

## 7. Formatting

```python
name = "Priyo"
score = 92.5

print(f"{name} scored {score:.1f}%")  # Priyo scored 92.5%
print("{} scored {:.1f}%".format(name, score))  # Priyo scored 92.5%
print("%s scored %.1f%%" % (name, score))  # Priyo scored 92.5%
```

## 8. Checking string content

```python
print("12345".isdigit())  # True
print("abc123".isalnum())  # True
print("abc".isalpha())  # True
print("   ".isspace())  # True
print("Title Case".istitle())  # True
```
