# List Operations

Lists are ordered, mutable, and allow duplicate elements.

```python
tasks = ["wash", "cook", "wash"]

# ordered -> position is preserved and always accessed the same way
print(tasks[0])  # wash

# mutable -> can reassign an element in place without creating a new list
tasks[1] = "clean"
print(tasks)  # ['wash', 'clean', 'wash']

# duplicates allowed -> "wash" appears twice and both are kept
print(tasks.count("wash"))  # 2
```

## 1. Creating and indexing

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])  # 10
print(numbers[-1])  # 50
print(numbers[1:4])  # [20, 30, 40]
print(numbers[::2])  # [10, 30, 50]  -> every 2nd element
```

## 2. Adding elements

```python
fruits = ["apple", "banana"]

fruits.append("cherry")
print(fruits)  # ['apple', 'banana', 'cherry']

fruits.insert(1, "kiwi")
print(fruits)  # ['apple', 'kiwi', 'banana', 'cherry']

fruits.extend(["date", "fig"])
print(fruits)  # ['apple', 'kiwi', 'banana', 'cherry', 'date', 'fig']
```

## 3. Removing elements

```python
fruits = ["apple", "kiwi", "banana", "cherry"]

fruits.remove("kiwi")  # removes first matching value
print(fruits)  # ['apple', 'banana', 'cherry']

popped = fruits.pop()  # removes & returns last item
print(popped, fruits)  # cherry ['apple', 'banana']

popped_at = fruits.pop(0)
print(popped_at, fruits)  # apple ['banana']

del fruits[0]
print(fruits)  # []
```

## 4. Sorting and reversing

```python
nums = [5, 2, 8, 1, 9]

nums.sort()
print(nums)  # [1, 2, 5, 8, 9]

nums.sort(reverse=True)
print(nums)  # [9, 8, 5, 2, 1]

nums.reverse()
print(nums)  # [1, 2, 5, 8, 9]

# sorted() returns a new list, leaving the original untouched
original = [3, 1, 2]
print(sorted(original), original)  # [1, 2, 3] [3, 1, 2]
```

## 5. List comprehensions

```python
squares = [n**2 for n in range(1, 6)]
print(squares)  # [1, 4, 9, 16, 25]

evens = [n for n in range(10) if n % 2 == 0]
print(evens)  # [0, 2, 4, 6, 8]

pairs = [(x, y) for x in range(2) for y in range(2)]
print(pairs)  # [(0, 0), (0, 1), (1, 0), (1, 1)]
```

## 6. Membership, length, aggregation

```python
nums = [4, 8, 15, 16, 23, 42]

print(15 in nums)  # True
print(len(nums))  # 6
print(sum(nums))  # 108
print(min(nums), max(nums))  # 4 42
print(nums.index(16))  # 3
print(nums.count(4))  # 1
```

## 7. Copying vs. referencing

Assignment copies the reference, not the data — both names point to the same list. Use `.copy()` (or `list(...)`, or slicing `[:]`) for an independent shallow copy.

```python
a = [1, 2, 3]
b = a  # b references the same list as a
b.append(4)
print(a)  # [1, 2, 3, 4]  -> a changed too!

c = a.copy()  # independent shallow copy
c.append(5)
print(a)  # [1, 2, 3, 4]     -> unaffected
print(c)  # [1, 2, 3, 4, 5]
```

## 8. Nested lists

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

print(matrix[1])  # [4, 5, 6]
print(matrix[1][2])  # 6

flattened = [value for row in matrix for value in row]
print(flattened)  # [1, 2, 3, 4, 5, 6, 7, 8, 9]
```
