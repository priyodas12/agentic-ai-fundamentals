# Tuple Operations

Tuples are ordered and, unlike lists, immutable — once created, their contents cannot change.

```python
coordinates = (10, 20, 30)

# ordered -> position is preserved and always accessed the same way
print(coordinates[0])  # 10
print(coordinates[1])  # 20

# immutable -> can't reassign an element after creation
try:
    coordinates[0] = 99
except TypeError as e:
    print(
        f"TypeError: {e}"
    )  # TypeError: 'tuple' object does not support item assignment
```

## 1. Creating and indexing

```python
point = (3, 4)
single = ("only one",)  # trailing comma required for a 1-element tuple
no_parens = 1, 2, 3  # parentheses are optional

print(point, type(point))  # (3, 4) <class 'tuple'>
print(single, type(single))  # ('only one',) <class 'tuple'>
print(no_parens, type(no_parens))  # (1, 2, 3) <class 'tuple'>
print(point[0])  # 3
print(point[-1])  # 4
```

## 2. Immutability

```python
point = (3, 4)
try:
    point[0] = 99
except TypeError as e:
    print(
        f"TypeError: {e}"
    )  # TypeError: 'tuple' object does not support item assignment
```

## 3. Packing and unpacking

```python
person = ("Alice", 30, "Metropolis")  # packing
name, age, city = person  # unpacking

print(name)  # Alice
print(age)  # 30
print(city)  # Metropolis

# star-unpacking to grab the "rest"
first, *rest = (1, 2, 3, 4)
print(first, rest)  # 1 [2, 3, 4]
```

## 4. Tuple methods, membership, slicing

```python
colors = ("red", "green", "blue", "green")

print(colors.count("green"))  # 2
print(colors.index("blue"))  # 2
print("red" in colors)  # True
print(colors[1:3])  # ('green', 'blue')
print(len(colors))  # 4
```

## 5. Concatenation and repetition

```python
a = (1, 2)
b = (3, 4)

print(a + b)  # (1, 2, 3, 4)
print(a * 3)  # (1, 2, 1, 2, 1, 2)
```

## 6. Why use a tuple over a list?

Tuples signal "this data shouldn't change" and are hashable (usable as dict keys / set members) as long as all their elements are hashable — lists are not.

```python
locations = {
    (51.5, -0.1): "London",
    (40.7, -74.0): "New York",
}
print(locations[(51.5, -0.1)])  # London

try:
    bad = {[1, 2]: "unhashable"}
except TypeError as e:
    print(f"TypeError: {e}")  # TypeError: unhashable type: 'list'
```

## 7. Nested tuples

```python
nested = ((1, 2), (3, 4), (5, 6))

print(nested[1])  # (3, 4)
print(nested[1][0])  # 3
```
