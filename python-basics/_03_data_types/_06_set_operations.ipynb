# Set Operations

Sets are unordered collections of unique, hashable elements.

```python
badges = {"gold", "silver", "gold", "bronze"}

# unordered -> no indexing by position
try:
    badges[0]
except TypeError as e:
    print(f"TypeError: {e}")  # TypeError: 'set' object is not subscriptable

# unique -> duplicate "gold" is dropped automatically
print(badges)  # {'bronze', 'gold', 'silver'}  (order not guaranteed)

# hashable elements only -> lists aren't hashable, so they can't be set members
try:
    {"gold", ["silver"]}
except TypeError as e:
    print(f"TypeError: {e}")  # TypeError: unhashable type: 'list'
```

## 1. Creating a set

```python
numbers = {1, 2, 3, 2, 1}  # duplicates are dropped automatically
empty_set = set()  # {} creates an empty DICT, not a set!
from_list = set([3, 1, 4, 1, 5])

print(numbers)  # {1, 2, 3}
print(empty_set)  # set()
print(from_list)  # {1, 3, 4, 5}
```

## 2. Adding and removing elements

```python
s = {1, 2, 3}

s.add(4)
print(s)  # {1, 2, 3, 4}

s.remove(2)  # raises KeyError if the element is missing
print(s)  # {1, 3, 4}

s.discard(99)  # no error even if the element is missing
print(s)  # {1, 3, 4}

popped = s.pop()  # removes & returns an arbitrary element
print(popped in {1, 3, 4})  # True
```

## 3. Set algebra: union, intersection, difference

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)  # union            -> {1, 2, 3, 4, 5, 6}
print(a & b)  # intersection      -> {3, 4}
print(a - b)  # difference        -> {1, 2}
print(b - a)  # difference        -> {5, 6}
print(a ^ b)  # symmetric diff    -> {1, 2, 5, 6}
```

## 4. Membership and subset/superset checks

```python
a = {1, 2, 3}
b = {1, 2, 3, 4, 5}

print(2 in a)  # True
print(a.issubset(b))  # True
print(b.issuperset(a))  # True
print(a.isdisjoint({9, 10}))  # True -> no common elements
```

## 5. Set comprehensions

```python
squares = {n**2 for n in range(-3, 4)}
print(squares)  # {0, 1, 4, 9}  -> negatives collapse into the same squares
```

## 6. `frozenset`: the immutable variant

```python
fs = frozenset([1, 2, 3])
print(fs, type(fs))  # frozenset({1, 2, 3}) <class 'frozenset'>

try:
    fs.add(4)
except AttributeError as e:
    print(f"AttributeError: {e}")  # 'frozenset' object has no attribute 'add'
```

## 7. Deduplicating a list (a very common use case)

```python
raw = [1, 2, 2, 3, 3, 3, 4]
unique_preserving_type = list(set(raw))

print(sorted(unique_preserving_type))  # [1, 2, 3, 4]
```
