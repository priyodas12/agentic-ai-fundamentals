# Dictionary (`dict`) Operations

Dicts store key-value pairs. Keys must be unique and hashable; since Python 3.7, insertion order is preserved.

```python
scores = {"alice": 90, "bob": 85}

# unique keys -> assigning to an existing key overwrites it, doesn't duplicate
scores["alice"] = 95
print(scores)  # {'alice': 95, 'bob': 85}

# keys must be hashable -> a list can't be used as a key
try:
    scores["alice"] = 100
except TypeError as e:
    print(f"TypeError: {e}")  # TypeError: unhashable type: 'list'

# insertion order preserved (Python 3.7+)
scores["carol"] = 70
print(list(scores.keys()))  # ['alice', 'bob', 'carol']
```

## 1. Creating and accessing

```python
person = {"name": "Alice", "age": 30, "city": "Metropolis"}

print(person["name"])  # Alice
print(person.get("age"))  # 30
print(person.get("country", "Unknown"))  # Unknown -> default when key is missing

try:
    person["country"]
except KeyError as e:
    print(f"KeyError: {e}")  # KeyError: 'country'
```

## 2. Adding, updating, removing

```python
person = {"name": "Alice", "age": 30}

person["city"] = "Metropolis"  # add a new key
person["age"] = 31  # update an existing key
print(person)  # {'name': 'Alice', 'age': 31, 'city': 'Metropolis'}

removed = person.pop("age")  # remove & return the value
print(removed, person)  # 31 {'name': 'Alice', 'city': 'Metropolis'}

del person["city"]
print(person)  # {'name': 'Alice'}
```

## 3. Iterating: keys, values, items

```python
person = {"name": "Alice", "age": 30, "city": "Metropolis"}

print(list(person.keys()))  # ['name', 'age', 'city']
print(list(person.values()))  # ['Alice', 30, 'Metropolis']
print(list(person.items()))  # [('name', 'Alice'), ('age', 30), ('city', 'Metropolis')]

for key, value in person.items():
    print(f"{key}: {value}")
# name: Alice
# age: 30
# city: Metropolis
```

## 4. `setdefault` and `update`

```python
counts = {}
for word in ["a", "b", "a", "c", "b", "a"]:
    counts[word] = counts.setdefault(word, 0) + 1
print(counts)  # {'a': 3, 'b': 2, 'c': 1}

defaults = {"theme": "light", "lang": "en"}
overrides = {"lang": "fr"}
defaults.update(overrides)
print(defaults)  # {'theme': 'light', 'lang': 'fr'}
```

## 5. Dict comprehensions

```python
squares = {n: n**2 for n in range(1, 6)}
print(squares)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

prices = {"apple": 1.5, "banana": 0.5, "cherry": 3.0}
discounted = {item: round(price * 0.9, 2) for item, price in prices.items()}
print(discounted)  # {'apple': 1.35, 'banana': 0.45, 'cherry': 2.7}
```

## 6. Merging dictionaries (Python 3.9+)

```python
a = {"x": 1, "y": 2}
b = {"y": 99, "z": 3}

merged = a | b  # right-hand side wins on key conflicts
print(merged)  # {'x': 1, 'y': 99, 'z': 3}

a |= b  # in-place merge
print(a)  # {'x': 1, 'y': 99, 'z': 3}
```

## 7. Nested dictionaries

```python
employees = {
    "e1": {"name": "Alice", "role": "Engineer"},
    "e2": {"name": "Bob", "role": "Designer"},
}

print(employees["e1"]["name"])  # Alice

for emp_id, info in employees.items():
    print(f"{emp_id}: {info['name']} ({info['role']})")
# e1: Alice (Engineer)
# e2: Bob (Designer)
```
