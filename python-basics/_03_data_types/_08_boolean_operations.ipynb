# Boolean (`bool`) Operations

`bool` is a subclass of `int` — `True` behaves as `1` and `False` as `0`.

```python
print(isinstance(True, int))  # True -> bool is a subclass of int
print(True == 1)  # True
print(False == 0)  # True
print(True + True)  # 2  -> arithmetic treats True as 1
```

## 1. Comparison operators produce booleans

```python
print(5 > 3)  # True
print(5 == 5)  # True
print(5 != 5)  # False
print(3 <= 2)  # False
```

## 2. Logical operators: `and`, `or`, `not`

```python
a, b = True, False

print(a and b)  # False
print(a or b)  # True
print(not a)  # False

# and/or return one of the actual operands, not just True/False
print(0 or "fallback")  # fallback
print("first" and "second")  # second
```

## 3. Short-circuit evaluation

`and`/`or` stop evaluating as soon as the result is determined — the right-hand side may never run.

```python
def noisy(label, value):
    print(f"evaluated {label}")
    return value


result = noisy("A", False) and noisy("B", True)
# evaluated A   -> B is never called since False short-circuits `and`
print(result)  # False

result = noisy("C", True) or noisy("D", True)
# evaluated C   -> D is never called since True short-circuits `or`
print(result)  # True
```

## 4. Truthy and falsy values

Every object has an implicit boolean value. Falsy: `0`, `0.0`, `""`, `[]`, `{}`, `()`, `set()`, `None`. Everything else is truthy.

```python
values = [0, 1, 0.0, "", "hi", [], [1], {}, {"a": 1}, None]

for v in values:
    print(repr(v), "->", bool(v))
# 0 -> False
# 1 -> True
# 0.0 -> False
# '' -> False
# 'hi' -> True
# [] -> False
# [1] -> True
# {} -> False
# {'a': 1} -> True
# None -> False
```

## 5. Booleans in arithmetic

```python
scores = [70, 85, 40, 90, 55]
passing = [s >= 60 for s in scores]

print(passing)  # [True, True, False, True, False]
print(sum(passing))  # 3  -> True counts as 1, False as 0
print(sum(passing) / len(scores))  # 0.6 -> pass rate
```

## 6. `all()` and `any()`

```python
nums = [2, 4, 6, 8]

print(all(n % 2 == 0 for n in nums))  # True  -> every element is even
print(any(n > 7 for n in nums))  # True  -> at least one element > 7
print(all([]))  # True  -> vacuously true for an empty iterable
print(any([]))  # False -> nothing to satisfy the condition
```
