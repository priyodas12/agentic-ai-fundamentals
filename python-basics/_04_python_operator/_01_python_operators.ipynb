# Python Operators

Goal: cover every category of Python operator with a runnable example — arithmetic, comparison, logical, assignment, bitwise, identity, and membership.

## 1. Arithmetic Operators

Used to perform mathematical operations: `+ - * / // % **`.

```python
a = 13
b = 4

print(a + b)  # 17  -> addition
print(a - b)  # 9   -> subtraction
print(a * b)  # 52  -> multiplication
print(a / b)  # 3.25 -> true division, always returns float
print(a // b)  # 3   -> floor division, rounds toward negative infinity
print(a % b)  # 1   -> modulus (remainder)
print(a**b)  # 28561 -> exponentiation
```

## 2. Comparison (Relational) Operators

Compare two values and always evaluate to a `bool`: `== != > < >= <=`.

```python
x = 10
y = 20

print(x == y)  # False -> equal to
print(x != y)  # True  -> not equal to
print(x > y)  # False -> greater than
print(x < y)  # True  -> less than
print(x >= 10)  # True  -> greater than or equal to
print(x <= 9)  # False -> less than or equal to
```

## 3. Logical Operators

Combine boolean expressions: `and`, `or`, `not`. `and`/`or` short-circuit and return one of the actual operands rather than a plain `True`/`False`.

```python
age = 25
has_id = True

print(age >= 18 and has_id)  # True  -> both conditions must be True
print(age < 18 or has_id)  # True  -> at least one condition is True
print(not has_id)  # False -> inverts the boolean

# and/or return an operand, not just True/False
print(0 or "fallback")  # fallback
print("first" and "second")  # second
```

## 4. Assignment Operators

`=` assigns a value; the augmented forms (`+= -= *= /= //= %= **=`) combine an operation with assignment in a single step.

```python
count = 10

count += 5  # count = count + 5
print(count)  # 15
count -= 3
print(count)  # 12
count *= 2
print(count)  # 24
count /= 4
print(count)  # 6.0  -> /= always produces a float
count //= 2
print(count)  # 3.0
count **= 3
print(count)  # 27.0
count %= 5
print(count)  # 2.0
```

## 5. Bitwise Operators

Operate on the binary representation of integers: `& | ^ ~ << >>`.

```python
a = 12  # 0b1100
b = 10  # 0b1010

print(bin(a), bin(b))  # 0b1100 0b1010
print(a & b)  # 8   -> 0b1000  AND: 1 only where both bits are 1
print(a | b)  # 14  -> 0b1110  OR: 1 where either bit is 1
print(a ^ b)  # 6   -> 0b0110  XOR: 1 where bits differ
print(~a)  # -13 -> NOT: inverts all bits (equivalent to -(a + 1))
print(a << 2)  # 48  -> left shift: multiplies by 2**2
print(a >> 2)  # 3   -> right shift: floor-divides by 2**2
```

## 6. Identity Operators

`is` / `is not` check whether two names point to the *same object* in memory — not whether their values are equal (that's `==`).

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)  # True  -> same contents
print(a is b)  # False -> different objects in memory
print(a is c)  # True  -> c references the exact same object as a
print(a is not b)  # True
```

## 7. Membership Operators

`in` / `not in` test whether a value exists within a sequence or collection.

```python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)  # True
print("mango" in fruits)  # False
print("mango" not in fruits)  # True
print("an" in "banana")  # True -> also works on strings (substring check)
```

## 8. Operator Precedence

When an expression mixes operators, Python evaluates higher-precedence operators first. Roughly, from highest to lowest: `**`, unary `+ - ~`, `* / // %`, `+ -`, bitwise shifts, `& ^ |`, comparisons, `not`, `and`, `or`.

```python
print(2 + 3 * 4)  # 14 -> multiplication before addition
print((2 + 3) * 4)  # 20 -> parentheses override precedence
print(2**3**2)  # 512 -> ** is right-associative: 2 ** (3 ** 2), not (2 ** 3) ** 2
print(not True or True)  # True -> `not` binds tighter than `or`: (not True) or True
```
