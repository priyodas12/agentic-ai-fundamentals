# Conditional Statements

Goal: cover `if`, `if-elif-else`, nested conditionals, the ternary expression, truthy/falsy checks, and structural pattern matching (`match`/`case`) with runnable examples.

## 1. `if` statement

Runs a block only when the condition is `True`.

```python
temperature = 35

if temperature > 30:
    print("It's hot outside")  # It's hot outside
```

## 2. `if-else`

Provides an alternative branch when the condition is `False`.

```python
age = 16

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote yet")  # You cannot vote yet
```

## 3. `if-elif-else`

Checks multiple conditions in order; only the first matching branch runs.

```python
score = 72

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(grade)  # C
```

## 4. Nested conditionals

An `if` block can contain another `if` inside it.

```python
name = ""
pin = ""

if name == "":
    if pin == "":
        print("Access granted")  # Access granted
    else:
        print("Wrong pin")
else:
    print("Unknown user")
```

## 5. Ternary (conditional) expression

A compact one-line form of `if-else` that evaluates to a value: `value_if_true if condition else value_if_false`.

```python
age = 20

status = "adult" if age >= 18 else "minor"
print(status)  # adult

# equivalent to:
if age >= 18:
    status = "adult"
else:
    status = "minor"
print(status)  # adult
```

## 6. Truthy and falsy values in conditions

`if` doesn't require a strict boolean — it evaluates any object's truthiness. Falsy: `0`, `0.0`, `""`, `[]`, `{}`, `()`, `None`. Everything else is truthy.

```python
items = []

if items:
    print("Cart has items")
else:
    print("Cart is empty")  # Cart is empty

name = ""
if not name:
    print("Name is missing")  # Name is missing
```

## 7. Combining conditions with logical operators

`and`, `or`, and `not` let a single `if` check multiple conditions.

```python
age = 25
has_ticket = True

if age >= 18 and has_ticket:
    print("Entry allowed")  # Entry allowed

is_weekend = False
is_holiday = True
if is_weekend or is_holiday:
    print("No work today")  # No work today
```

## 8. `match` / `case` — structural pattern matching (Python 3.10+)

An alternative to long `if-elif` chains when matching a value against several fixed patterns.

```python
def describe_status(code):
    match code:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case 500 | 502 | 503:  # matches any of these
            return "Server Error"
        case _:  # wildcard, like `else`
            return "Unknown Status"


print(describe_status(200))  # OK
print(describe_status(503))  # Server Error
print(describe_status(301))  # Unknown Status
```

## 9. No `switch` statement: dict-based dispatch

Python has never had a `switch`/`case` keyword. Before `match`/`case` (3.10+) existed, the common workaround was a dictionary mapping each case to a value or function — an O(1) lookup that also avoids a long `if-elif` chain.

```python
# simple case: mapping straight to values
status_names = {200: "OK", 404: "Not Found", 500: "Server Error"}

print(status_names.get(200))  # OK
print(
    status_names.get(301, "Unknown Status")
)  # Unknown Status -> default for missing keys


# richer case: mapping to functions, called only when matched (lazy dispatch)
def handle_get():
    return "Handling GET request"


def handle_post():
    return "Handling POST request"


def handle_unknown():
    return "Method not allowed"


dispatch = {
    "GET": handle_get,
    "POST": handle_post,
}

method = "POST"
result = dispatch.get(method, handle_unknown)()  # look up the function, then call it
print(result)  # Handling POST request

method = "DELETE"
result = dispatch.get(method, handle_unknown)()
print(result)  # Method not allowed
```
