# Python Loops

Goal: browse across the different types of loops in Python — `for`, `while`, iteration helpers (`range`, `enumerate`, `zip`), loop control (`break`/`continue`), the loop `else` clause, and nested loops.

## 1. `for` loop with `range()`

`range(start, stop, step)` generates a sequence of numbers without building a full list in memory.

```python
for i in range(5):
    print(i)
# 0
# 1
# 2
# 3
# 4

for i in range(2, 10, 2):  # start=2, stop=10 (exclusive), step=2
    print(i)
# 2
# 4
# 6
# 8
```

## 2. `while` loop

Repeats as long as its condition stays `True`. The condition is checked before every iteration.

```python
count = 0
while count < 5:
    print(count)
    count += 1
# 0
# 1
# 2
# 3
# 4
```

## 3. Iterating over collections

`for` can iterate directly over any iterable — lists, strings, tuples, dicts, sets.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
# apple
# banana
# cherry

for letter in "abc":
    print(letter)
# a
# b
# c

scores = {"alice": 90, "bob": 85}
for key, value in scores.items():  # iterating key-value pairs
    print(key, value)
# alice 90
# bob 85
```

## 4. `enumerate()`

Pairs each element with its index — no need to manage a separate counter variable.

```python
colors = ["red", "green", "blue"]

for index, color in enumerate(colors):
    print(index, color)
# 0 red
# 1 green
# 2 blue

for index, color in enumerate(colors, start=1):  # custom starting index
    print(index, color)
# 1 red
# 2 green
# 3 blue
```

## 5. `zip()`

Iterates over multiple iterables in parallel, pairing up elements by position.

```python
names = ["Alice", "Bob", "Carol"]
ages = [30, 25, 35]

for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
# Alice is 30 years old
# Bob is 25 years old
# Carol is 35 years old

# zip() stops at the shortest iterable
print(list(zip([1, 2, 3], ["a", "b"])))  # [(1, 'a'), (2, 'b')]
```

## 6. `break` and `continue`

`break` exits the loop entirely. `continue` skips the rest of the current iteration and moves on to the next one.

```python
for n in range(10):
    if n == 5:
        break  # stop the loop entirely once n reaches 5
    print(n)
# 0 1 2 3 4

for n in range(6):
    if n % 2 == 0:
        continue  # skip even numbers
    print(n)
# 1 3 5
```

## 7. Loop `else` clause

The `else` block of a `for`/`while` loop runs only if the loop completed *without* hitting a `break`. Useful for "search and report if nothing was found" patterns.

```python
numbers = [1, 3, 5, 7, 9]

for n in numbers:
    if n % 2 == 0:
        print("Found an even number")
        break
else:
    print("No even numbers found")  # runs because break never happened

for n in [1, 3, 4, 7]:
    if n % 2 == 0:
        print("Found an even number")  # Found an even number
        break
else:
    print("No even numbers found")  # skipped, because break happened
```

## 8. Nested loops

A loop can contain another loop; the inner loop completes fully for every single iteration of the outer loop.

```python
for row in range(3):
    for col in range(3):
        print(f"({row}, {col})", end=" ")
    print()  # newline after each row
# (0, 0) (0, 1) (0, 2)
# (1, 0) (1, 1) (1, 2)
# (2, 0) (2, 1) (2, 2)
```

## 9. Infinite loop with a sentinel `break`

`while True` loops forever unless something inside explicitly `break`s out — a common pattern for "keep going until a condition is met", e.g. reading input.

```python
values = iter([4, 8, 15, 16, 23, 42])  # stand-in for something like user input
total = 0

while True:
    try:
        value = next(values)
    except StopIteration:
        break  # no more values -> exit the infinite loop
    total += value

print(total)  # 108
```

## 10. List/dict/set comprehensions as a loop shorthand

A comprehension is a compact `for` loop that builds a new collection in one line.

```python
squares = [n**2 for n in range(1, 6)]  # equivalent to a for-loop with .append()
print(squares)  # [1, 4, 9, 16, 25]

even_squares = {
    n: n**2 for n in range(1, 6) if n % 2 == 0
}  # dict comprehension with a filter
print(even_squares)  # {2: 4, 4: 16}
```
