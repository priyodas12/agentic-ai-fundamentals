# Python Standard Library

Goal: browse a broad tour of the Python standard library — `os`, `sys`, `datetime`, `time`, `math`, `random`, `collections`, `itertools`, `functools`, `re`, `json`, `pathlib`, `statistics`, `enum`, `hashlib`/`uuid`, `textwrap`, and `copy`. Each section shows a few of the most commonly reached-for functions in that module.

## 1. `os` — operating system interfaces

```python
import os

print(os.getcwd())              # current working directory
print(os.name)                  # 'posix' on Linux/Mac, 'nt' on Windows
print(os.path.join("a", "b", "c.txt"))   # builds a path in an OS-appropriate way
print(os.path.split("/home/user/file.txt"))
print(os.path.splitext("archive.tar.gz"))
print(list(os.environ.keys())[:3])  # a peek at available environment variable names
```

## 2. `sys` — interpreter internals

```python
import sys

print(sys.version_info[:3])   # (major, minor, micro)
print(sys.platform)
print(type(sys.path))         # list of directories Python searches for imports
print(sys.maxsize)
```

## 3. `datetime` — dates, times, and deltas

```python
from datetime import datetime, date, timedelta

now = datetime(2026, 8, 14, 9, 30, 0)  # fixed moment, so output is reproducible
print(now)
print(now.strftime("%Y-%m-%d %H:%M"))
print(now.year, now.month, now.day)

tomorrow = now + timedelta(days=1)
print(tomorrow)

d1 = date(2026, 1, 1)
d2 = date(2026, 8, 14)
print((d2 - d1).days)
```

## 4. `time` — timing and epoch-based time

```python
import time

start = time.perf_counter()
total = sum(range(1_000_000))  # something to measure
elapsed = time.perf_counter() - start

print(total)
print(elapsed >= 0)
print(time.strftime("%H:%M:%S", time.gmtime(3661)))  # format a fixed epoch offset
```

## 5. `math` — numeric functions and constants

```python
import math

print(math.sqrt(16))
print(math.floor(4.7), math.ceil(4.2))
print(math.pi, math.e)
print(math.factorial(5))
print(math.gcd(48, 18))
print(round(math.log(math.e)))
```

## 6. `random` — pseudo-random values

`random.seed(...)` makes the sequence reproducible — useful for demos and tests.

```python
import random

random.seed(42)  # fixed seed -> reproducible output

print(random.randint(1, 100))
print(random.choice(["red", "green", "blue"]))
print(random.sample(range(1, 50), 6))
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(numbers)
```

## 7. `collections` — specialized container types

```python
from collections import namedtuple, Counter, defaultdict, deque

Point = namedtuple("Point", ["x", "y"])
p = Point(3, 4)
print(p, p.x, p.y)

counts = Counter("mississippi")
print(counts)
print(counts.most_common(2))

dd = defaultdict(list)
dd["fruits"].append("apple")
dd["fruits"].append("banana")
print(dict(dd))

dq = deque([1, 2, 3])
dq.appendleft(0)
dq.append(4)
print(dq)
```

## 8. `itertools` — building blocks for iterators

```python
import itertools

print(list(itertools.chain([1, 2], [3, 4], [5])))
print(list(itertools.combinations("ABC", 2)))
print(list(itertools.permutations("AB")))

for key, group in itertools.groupby(sorted(["a", "ab", "b", "bc", "c"], key=len), key=len):
    print(key, list(group))

print(list(itertools.islice(itertools.cycle([1, 2, 3]), 7)))
```

## 9. `functools` — higher-order function helpers

```python
from functools import reduce, lru_cache, partial, wraps

print(reduce(lambda a, b: a + b, [1, 2, 3, 4]))

@lru_cache(maxsize=None)
def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)

print(fib(20))

add = lambda a, b: a + b
add5 = partial(add, 5)
print(add5(10))

def logger(func):
    @wraps(func)  # preserves func.__name__ / docstring on the wrapper
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@logger
def greet():
    """Say hello."""
    return "hi"

print(greet.__name__, greet.__doc__)
```

## 10. `re` — regular expressions

```python
import re

text = "Contact: alice@example.com or bob@test.org"

print(re.search(r"\w+@\w+\.\w+", text).group())
print(re.findall(r"\w+@\w+\.\w+", text))
print(re.sub(r"@\w+\.\w+", "@[hidden]", text))

match = re.match(r"(\w+): (.+)", "Contact: info here")
print(match.groups())
```

## 11. `json` — serialization to/from JSON text

```python
import json

data = {"name": "Alice", "active": True, "scores": [90, 85, 88]}

as_text = json.dumps(data, indent=2, sort_keys=True)
print(as_text)

back = json.loads(as_text)
print(back == data)
```

## 12. `pathlib` — object-oriented filesystem paths

```python
from pathlib import Path

p = Path("static") / "notes.txt"
print(p)
print(p.parent, p.name, p.suffix)

here = Path(".")
print(here.resolve().name)  # name of the current directory
```

## 13. `statistics` — basic descriptive statistics

```python
import statistics

data = [4, 8, 15, 16, 23, 42]

print(statistics.mean(data))
print(statistics.median(data))
print(round(statistics.stdev(data), 2))
print(statistics.mode([1, 1, 2, 3]))
```

## 14. `enum` — named constant groups

```python
from enum import Enum, auto

class Color(Enum):
    RED = auto()
    GREEN = auto()
    BLUE = auto()

print(Color.RED)
print(Color.RED.name, Color.RED.value)
print(list(Color))
print(Color.RED == Color.RED, Color.RED == Color.GREEN)
```

## 15. `hashlib` and `uuid` — hashing and unique identifiers

```python
import hashlib
import uuid

digest = hashlib.sha256(b"hello world").hexdigest()
print(digest)

id1 = uuid.uuid4()
print(len(str(id1)))  # standard uuid4 string is always 36 chars
print(str(id1).count("-"))
```

## 16. `textwrap` — wrapping and shortening text

```python
import textwrap

text = "Python's standard library ships with a huge set of ready-to-use modules."

print(textwrap.fill(text, width=30))
print(textwrap.shorten(text, width=40, placeholder="..."))
```

## 17. `copy` — shallow vs. deep copies

A shallow copy duplicates the outer container but still shares nested mutable objects with the original. A deep copy recursively duplicates everything, so the two are fully independent.

```python
import copy

original = {"items": [1, 2, 3], "name": "cart"}

shallow = copy.copy(original)
shallow["items"].append(4)  # shallow copy shares the nested list
print(original["items"])

deep = copy.deepcopy(original)
deep["items"].append(99)  # deep copy has its own independent nested list
print(original["items"], deep["items"])
```
